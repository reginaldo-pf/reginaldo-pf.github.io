## 1. Módulo de Taxonomia de Atributos e Metamodelo do MER (Peter Chen)

```python
from enum import Enum
from dataclasses import dataclass, field
from typing import List, Optional, Callable, Dict, Any

class TipoAtributo(Enum):
    IDENTIFICADOR = "Identificador (Chave Primária - Sublinhado no DER)"
    SIMPLES = "Simples / Atômico (Elipse Sólida)"
    COMPOSTO = "Composto (Elipse Ramificada)"
    MULTIVALORADO = "Multivalorado (Elipse Dupla Concêntrica)"
    DERIVADO = "Derivado (Elipse Tracejada)"

@dataclass
class DefinicaoAtributo:
    nome: str
    tipo: TipoAtributo
    subatributos: List["DefinicaoAtributo"] = field(default_factory=list)
    regra_derivacao: Optional[str] = None
    funcao_calculo: Optional[Callable[[Dict[str, Any]], Any]] = None

@dataclass
class EsquemaEntidade:
    nome: str
    atributos: List[DefinicaoAtributo]

    def obter_chave(self) -> Optional[DefinicaoAtributo]:
        for a in self.atributos:
            if a.tipo == TipoAtributo.IDENTIFICADOR:
                return a
        return None

    def gerar_mermaid(self) -> str:
        linhas = ["flowchart TD"]
        e_id = f"ent_{self.nome}"
        linhas.append(f'    {e_id}["{self.nome}"]:::entidade')
        for i, a in enumerate(self.atributos):
            a_id = f"attr_{self.nome}_{i}"
            if a.tipo == TipoAtributo.IDENTIFICADOR:
                linhas.append(f'    {a_id}(["<u>{a.nome}</u>"]):::chave')
                linhas.append(f"    {e_id} --- {a_id}")
            elif a.tipo == TipoAtributo.SIMPLES:
                linhas.append(f'    {a_id}(["{a.nome}"]):::simples')
                linhas.append(f"    {e_id} --- {a_id}")
            elif a.tipo == TipoAtributo.MULTIVALORADO:
                linhas.append(f'    {a_id}([["{a.nome}"]]):::multivalorado')
                linhas.append(f"    {e_id} --- {a_id}")
            elif a.tipo == TipoAtributo.DERIVADO:
                linhas.append(f"    {a_id}-.- {e_id}")
                linhas.append(f'    {a_id}(["{a.nome}"]):::derivado')
            elif a.tipo == TipoAtributo.COMPOSTO:
                linhas.append(f'    {a_id}(["{a.nome}"]):::composto')
                linhas.append(f"    {e_id} --- {a_id}")
                for j, sub in enumerate(a.subatributos):
                    sub_id = f"{a_id}_sub_{j}"
                    linhas.append(f'    {sub_id}(["{sub.nome}"]):::simples')
                    linhas.append(f"    {a_id} --- {sub_id}")
        linhas.append("    classDef entidade fill:#059669,stroke:#047857,stroke-width:2px,color:#ffffff,font-weight:bold;")
        linhas.append("    classDef chave fill:#e0f2fe,stroke:#0284c7,stroke-width:2px,color:#0369a1,font-weight:bold;")
        linhas.append("    classDef simples fill:#f1f5f9,stroke:#64748b,stroke-width:1.5px,color:#1e293b;")
        linhas.append("    classDef composto fill:#fef3c7,stroke:#d97706,stroke-width:1.5px,color:#92400e;")
        linhas.append("    classDef multivalorado fill:#fce7f3,stroke:#db2777,stroke-width:3px,color:#9d174d;")
        linhas.append("    classDef derivado fill:#ede9fe,stroke:#7c3aed,stroke-dasharray: 5 5,stroke-width:2px,color:#5b21b6;")
        return "\n".join(linhas)

print("Módulo de Metamodelo do MER carregado com sucesso!")
```

## 2. Diagrama da Questão 9: Arquitetura ANSI/SPARC de 3 Níveis
A célula a seguir gera o código Mermaid para você verificar a estrutura esquemática completa da arquitetura exigida na Questão 9.

```python
diagrama_ansisparc = """flowchart TB
    subgraph NIVEL_EXTERNO ["NÍVEL EXTERNO (Visões de Usuários e Aplicações)"]
        direction LR
        V1["Visão 1: Operador do NOC<br/>(Topologia e Alarmes)"]
        V2["Visão 2: Gestor de TI<br/>(Inventário e Custos)"]
        V3["Visão 3: Faturamento do Provedor<br/>(Consumo de Banda e Tráfego)"]
    end

    subgraph MAP_1 ["Mapeamento Externo-Conceitual"]
        M1["Assegura Independência Lógica"]
    end

    subgraph NIVEL_CONCEITUAL ["NÍVEL CONCEITUAL (Esquema Global Unificado)"]
        C1["Modelo Lógico Global<br/>Entidades: POP, ENLACE, SERVIDOR, ROTEADOR<br/>Restrições de Integridade & Regras de Negócio"]
    end

    subgraph MAP_2 ["Mapeamento Conceitual-Interno"]
        M2["Assegura Independência Física"]
    end

    subgraph NIVEL_INTERNO ["NÍVEL INTERNO (Esquema Físico de Armazenamento)"]
        direction LR
        I1["Arquivos de Disco"]
        I2["Índices B-Tree e Hash"]
        I3["Buffer Pool e Blocos Físicos"]
    end

    subgraph CATALOGO ["CATÁLOGO DO SISTEMA (Dicionário de Metadados)"]
        CAT["Metadados Centralizados<br/>Consultado por: Otimizador CBO e Módulo de Segurança"]
    end

    NIVEL_EXTERNO <--> MAP_1
    MAP_1 <--> NIVEL_CONCEITUAL
    NIVEL_CONCEITUAL <--> MAP_2
    MAP_2 <--> NIVEL_INTERNO

    CAT -. Metadados de Visões .- NIVEL_EXTERNO
    CAT -. Esquema Lógico .- NIVEL_CONCEITUAL
    CAT -. Métodos de Acesso .- NIVEL_INTERNO"""

print("=== Código Mermaid da Questão 9 (ANSI/SPARC) ===")
print(diagrama_ansisparc)
```

## 3. Diagrama da Questão 13: Entidade SERVIDOR_VIRTUALIZADO (Peter Chen)
A célula a seguir instancia o esquema formal da Questão 13 e exporta o código do DER na notação canônica de Peter Chen.

```python
esquema_q13 = EsquemaEntidade(
    nome="SERVIDOR_VIRTUALIZADO",
    atributos=[
        DefinicaoAtributo(nome="uuid_maquina", tipo=TipoAtributo.IDENTIFICADOR),
        DefinicaoAtributo(nome="nome_virtual", tipo=TipoAtributo.SIMPLES),
        DefinicaoAtributo(nome="memoria_ram_gb", tipo=TipoAtributo.SIMPLES),
        DefinicaoAtributo(
            nome="alocacao_cluster",
            tipo=TipoAtributo.COMPOSTO,
            subatributos=[
                DefinicaoAtributo(nome="nome_datacenter", tipo=TipoAtributo.SIMPLES),
                DefinicaoAtributo(nome="host_fisico_id", tipo=TipoAtributo.SIMPLES),
                DefinicaoAtributo(nome="storage_pool", tipo=TipoAtributo.SIMPLES),
            ]
        ),
        DefinicaoAtributo(nome="enderecos_ip_interfaces", tipo=TipoAtributo.MULTIVALORADO),
        DefinicaoAtributo(
            nome="custo_estimado_mensal",
            tipo=TipoAtributo.DERIVADO,
            regra_derivacao="(memoria_ram_gb * 15.0) + (storage_capacidade_gb * 0.5)",
            funcao_calculo=lambda d: (d.get("memoria_ram_gb", 0) * 15.0) + (d.get("storage_capacidade_gb", 0) * 0.5)
        )
    ]
)

print("=== Código Mermaid do DER da Questão 13 ===")
print(esquema_q13.gerar_mermaid())
```
