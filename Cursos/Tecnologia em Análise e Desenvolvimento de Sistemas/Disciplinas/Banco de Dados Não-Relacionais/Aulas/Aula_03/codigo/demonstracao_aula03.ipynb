## ⚡ 3. Prática com Redis (redis-py)
Exploramos as 5 estruturas centrais do Redis:
- **Strings**: Tokens de sessão com tempo de vida (`EXPIRE` / `TTL`) e contadores atômicos (`INCR`).
- **Hashes**: Representação de entidades e carrinhos de compras (`HSET`, `HGETALL`).
- **Lists**: Filas FIFO assíncronas para processamento de pedidos em background (`RPUSH`, `LPOP`).
- **Sets**: Coleções únicas de tags (`SADD`, `SMEMBERS`).
- **Sorted Sets (ZSets)**: Placares e rankings de produtos mais vendidos (`ZADD`, `ZREVRANGE`).

```python
# Demonstração conceitual das estruturas de dados do Redis em Python
print("Estruturas Chave do Redis:")
print("1. Strings:     SET session:token <jwt> EX 120")
print("2. Contadores:  INCR views:NOT-DEL-01")
print("3. Hashes:      HSET cart:user_42 NOT-DEL-01 1 MOU-LOG-02 2")
print("4. Lists:       RPUSH fila:pedidos PED-1001 -> LPOP fila:pedidos")
print("5. Sorted Sets: ZADD ranking 45 MOU-LOG-02 12 NOT-DEL-01")
```
