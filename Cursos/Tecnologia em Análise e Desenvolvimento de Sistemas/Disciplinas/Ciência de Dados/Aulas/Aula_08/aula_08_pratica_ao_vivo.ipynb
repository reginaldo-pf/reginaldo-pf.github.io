### 2. Exemplo 1: A Média como "Equalizador" (*Smoother*)
**Cenário dos Slides:** 4 pessoas possuem nas suas carteiras os valores \$2, \$3, \$3 e \$9.

```python
# Colecao not_symmetric dos slides
not_symmetric = np.array([2, 3, 3, 9])

total_dinheiro = np.sum(not_symmetric)
media_dinheiro = np.mean(not_symmetric)

print(f"Total arrecadado no fundo comum: R$ {total_dinheiro:.2f}")
print(f"Valor redistribuido igualmente (Média): R$ {media_dinheiro:.2f}")
```

### 3. Exemplo 2: Proporções são Médias (Variáveis Booleanas em ADS)
**Conceito:** Se uma coleção é formada apenas por `0`s e `1`s (ou `False` e `True`), a média é exatamente a **proporção de sucessos**.

```python
# Array binario [1, 1, 1, 0]
zero_one = np.array([1, 1, 1, 0])
print("Soma dos 1s:", np.sum(zero_one))
print("Media (Proporcao de 1s):", np.mean(zero_one))

# Array booleano [True, True, True, False]
bool_array = np.array([True, True, True, False])
print("Media de booleanos:", np.mean(bool_array))
```

#### 💻 Aplicação Prática em ADS: Log de Cliques em Web App (CTR)
Vamos simular 1.000 requisições de um sistema web onde monitoramos se o usuário clicou em um anúncio (`1`) ou não (`0`).

```python
np.random.seed(42)
cliques = np.random.choice([0, 1], size=1000, p=[0.85, 0.15]) # 15% de probabilidade de clique

df_web = pd.DataFrame({
    "usuario_id": range(1000),
    "clicou": cliques
})

ctr = df_web["clicou"].mean()
print(f"Taxa de Clique (Click-Through Rate - CTR): {ctr:.2%}")
print(f"Total de cliques no sistema: {df_web['clicou'].sum()}")
```

### 4. Exemplo 3: A Média Depende Apenas da Distribuição
**Conceito:** Duas coleções com a mesma proporção relativa de valores possuem exatamente a mesma média, independente do número de elementos.

```python
colecao_A = np.array([2, 3, 3, 9])
colecao_B = np.array([2, 2, 3, 3, 3, 3, 9, 9])

print("Tamanho de A:", len(colecao_A), "| Media de A:", np.mean(colecao_A))
print("Tamanho de B:", len(colecao_B), "| Media de B:", np.mean(colecao_B))
```

---
### 5. Exemplo 4: Visualizando a Média como Centro de Gravidade (*Fulcrum*)
Vamos plotar o histograma da coleção `{2, 3, 3, 9}` e marcar o ponto de equilíbrio no valor **4.25**.

```python
fig, ax = plt.subplots(figsize=(8, 4))

bins = np.arange(1.5, 10.5, 1)
counts, _, patches = ax.hist(colecao_A, bins=bins, density=True, color="#0EA5E9", edgecolor="#1E293B", alpha=0.7, rwidth=0.9)

media = np.mean(colecao_A)

# Plot do Ponto de Apoio (Fulcrum)
ax.scatter(media, 0, color="#F59E0B", s=250, marker="^", zorder=5, label=f"Média (Centro de Gravidade) = {media}")
ax.axhline(0, color="black", linewidth=2)

ax.set_xticks(range(2, 10))
ax.set_title("Histograma de {2, 3, 3, 9} e seu Ponto de Apoio", fontsize=14, fontweight="bold")
ax.set_xlabel("Valores")
ax.set_ylabel("Proporção (Frequência Relativa)")
ax.legend(loc="upper right", frameon=True)
plt.savefig("histograma_ponto_apoio.png", dpi=300, bbox_inches="tight")
plt.show()
```

### 6. Exemplo 5: Média vs. Mediana e o Mito do "Abaixo da Média"
Compare a distribuição simétrica `{2, 3, 3, 4}` com a assimétrica `{2, 3, 3, 9}`.

```python
simetrica = np.array([2, 3, 3, 4])
nao_simetrica = np.array([2, 3, 3, 9])

df_comp = pd.DataFrame({
    "Conjunto": ["Simétrico {2,3,3,4}", "Assimétrico {2,3,3,9}"],
    "Média": [np.mean(simetrica), np.mean(nao_simetrica)],
    "Mediana": [np.median(simetrica), np.median(nao_simetrica)],
    "% dados < Média": [
        f"{np.mean(simetrica < np.mean(simetrica)):.0%}",
        f"{np.mean(nao_simetrica < np.mean(nao_simetrica)):.0%}"
    ]
})
display(df_comp)
```

### 7. Estudo de Caso Real: Salários e Compensações (San Francisco 2015)
Vamos carregar ou simular o conjunto de dados de salários de San Francisco para observar o comportamento da **Média** e da **Mediana** sob assimetria acentuada (*Right Skewed*).

```python
np.random.seed(42)
# Simulacao fiel da distribuicao de salarios de SF 2015 (Compensacao Total > $10.000)
base_salarios = np.random.lognormal(mean=11.5, sigma=0.5, size=5000)
salarios_sf = base_salarios[(base_salarios >= 10000) & (base_salarios <= 700000)]

media_sf = np.mean(salarios_sf)
mediana_sf = np.median(salarios_sf)

print(f"Mediana do Salário: R$ {mediana_sf:,.2f}  (Salário Típico)")
print(f"Média do Salário:   R$ {media_sf:,.2f}  (Puxada pela cauda longa)")
print(f"Diferença:          R$ {(media_sf - mediana_sf):,.2f}")
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

sns.histplot(salarios_sf, bins=40, kde=True, ax=ax, color="#0EA5E9", edgecolor="black")

ax.axvline(mediana_sf, color="#E11D48", linestyle="--", linewidth=2.5, label=f"Mediana: R$ {mediana_sf:,.0f}")
ax.axvline(media_sf, color="#F59E0B", linestyle="-", linewidth=2.5, label=f"Média: R$ {media_sf:,.0f}")

ax.set_title("Distribuição de Salários (San Francisco 2015) - Assimetria à Direita", fontsize=14, fontweight="bold")
ax.set_xlabel("Compensação Total (R\$)")
ax.set_ylabel("Frequência")
ax.legend(fontsize=12)
plt.savefig("salarios_sf_assimetria.png", dpi=300, bbox_inches="tight")
plt.show()
```

### 8. Desafio Prático de ADS (Live Coding em Sala!)
**Cenário em Engenharia de Software:**
Você é o engenheiro de dados responsável pelo monitoramento de performance de uma API de pagamento.
Abaixo temos o log do tempo de resposta (latência em milissegundos) de 1.000 requisições.

```python
np.random.seed(123)
# 95% das requisicoes respondem rapido (50ms - 150ms), 5% sofrem timeout/delay alto (2000ms - 5000ms)
latencias_normais = np.random.normal(loc=100, scale=20, size=950)
latencias_lentas = np.random.uniform(low=2000, high=5000, size=50)
latencias_api = np.concatenate([latencias_normais, latencias_lentas])

print("Primeiras 10 latencias:", latencias_api[:10].round(1))
```

**PERGUNTA PARA A TURMA:**
1. Calcule a média e a mediana do tempo de resposta da API.
2. Qual métrica você colocaria no SLA do painel de monitoramento do sistema para refletir a experiência típica do usuário?
3. Qual métrica você usaria para projetar o consumo total de banda/processador do servidor?

```python
# RESPOSTA DO EXERCÍCIO DA TURMA
media_lat = np.mean(latencias_api)
mediana_lat = np.median(latencias_api)

print(f"Média de Latência:   {media_lat:.2f} ms  (Afetada drasticamente pelos 5% de requisições lentas)")
print(f"Mediana de Latência: {mediana_lat:.2f} ms  (Reflete a latência do usuário típico)")
```
