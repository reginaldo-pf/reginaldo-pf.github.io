### 2. Módulo 1: Revisão Essencial de Probabilidade para Ciência de Dados
**Conceitos:** Espaço Amostral ($\Omega$), Eventos Equiprováveis, Regra da Adição (Eventos Mutuamente Exclusivos vs. Não-Exclusivos) e Regra da Multiplicação (Probabilidade Condicional $P(B|A)$ e Independência).

```python
# 1. Experimento: Lancamento de 2 Moedas (HH, HT, TH, TT)
np.random.seed(42)
m1 = np.random.choice(['H', 'T'], size=10000)
m2 = np.random.choice(['H', 'T'], size=10000)

# P(Pelo menos uma Cara) = 3/4 = 75%
prob_pelo_menos_uma_cara = np.mean((m1 == 'H') | (m2 == 'H'))
print(f"P(Pelo menos uma Cara): {prob_pelo_menos_uma_cara:.2%}")

# 2. Regra da Adição: Eventos Mutuamente Exclusivos
# P(17 anos) = 1.4%, P(18 anos) = 1.5% -> P(17 ou 18 anos) = 1.4% + 1.5% = 2.9%
p_17, p_18 = 0.014, 0.015
print(f"P(17 ou 18 anos): {p_17 + p_18:.1%}")

# 3. Probabilidade Condicional e Amostragem Sem Reposição
# Sorteando 2 alunos de {X, Y, Z} sem reposição. P(Y primeiro E X segundo) = (1/3) * (1/2) = 1/6
p_yx = (1/3) * (1/2)
print(f"P(Y 1º e X 2º sem reposição): {p_yx:.4f} (1/6)")
```

---
### 3. Módulo 2: Amostragem Aleatória em Python (Com e Sem Reposição)
**Conceito:** A amostragem **com reposição** gera eventos independentes, enquanto a amostragem **sem reposição** altera o espaço amostral a cada passo (eventos dependentes).

```python
alunos = np.array(['Aluno_A', 'Aluno_B', 'Aluno_C', 'Aluno_D', 'Aluno_E'])

# Sorteio COM reposicao (replace=True)
amostra_com = np.random.choice(alunos, size=3, replace=True)
print("Amostra COM Reposição:", amostra_com)

# Sorteio SEM reposicao (replace=False)
amostra_sem = np.random.choice(alunos, size=3, replace=False)
print("Amostra SEM Reposição:", amostra_sem)

# Amostragem aleatoria em DataFrames com df.sample()
df_users = pd.DataFrame({
    'id': range(1, 101),
    'ativo': np.random.choice([0, 1], size=100, p=[0.3, 0.7])
})
display(df_users.sample(n=5, random_state=42))
```

### 4. Módulo 3: A Média e suas Propriedades Fundamentais
**Exemplo:** Coleção `{2, 3, 3, 9}` (Média = 4.25, Ponto de Equilíbrio / Smoother).

```python
not_symmetric = np.array([2, 3, 3, 9])
media = np.mean(not_symmetric)
mediana = np.median(not_symmetric)

print(f"Média: {media:.2f} | Mediana: {mediana:.2f}")
print(f"Proporção de dados < Média: {np.mean(not_symmetric < media):.0%}")

# Proporção em DataFrames (df['ativo'].mean())
taxa_ativos = df_users['ativo'].mean()
print(f"Taxa de Usuários Ativos (Proporção): {taxa_ativos:.2%}")
```
